```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Jika menggunakan NumPy array, harga * 0.9 akan mengalikan setiap elemen harga dengan 0.9 sehingga menghasilkan harga setelah diskon 10%. Jika menggunakan list Python biasa, [5000,7000,3000] * 0.9 akan menghasilkan TypeError karena list tidak dapat dikalikan dengan bilangan desimal. Pada list, operator * digunakan untuk mengulang isi list dengan bilangan bulat, sedangkan NumPy array mendukung operasi matematika secara langsung pada setiap elemennya.

```python
import pandas as pd
data_kantin = {
'menu': ['Nasi Goreng', 'Es Teh', 'Mie Ayam', 'Es Teh', None],
'harga': [12000, 4000, 10000, 4000, 8000],
'terjual': [23, 40, None, 35, 18]
}
df = pd.DataFrame(data_kantin)
print(df)
```

Kolom yang memiliki data kosong (None) adalah kolom menu dan terjual. Data kosong perlu dibersihkan sebelum dianalisis karena dapat menyebabkan hasil analisis menjadi tidak akurat. Misalnya, data jumlah barang terjual yang kosong dapat memengaruhi perhitungan rata-rata dan total penjualan. Data menu yang kosong juga membuat kita sulit mengetahui jenis makanan yang terkait dengan data tersebut.

```python
df = pd.read_csv('data_kantin.csv')
print(df.head()) # 5 baris pertama
print(df.info()) # tipe data & jumlah non-null tiap kolom
print(df.describe()) # statistik ringkas kolom numerik
print(df.shape) # jumlah (baris, kolom)
```

Kolom yang jumlah non-null-nya lebih sedikit dari jumlah baris total adalah kolom yang memiliki data kosong (missing value). Artinya, terdapat beberapa baris yang tidak memiliki nilai pada kolom tersebut sehingga datanya perlu diperiksa atau ditangani.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

Kolom terjual diisi dengan 0 karena nilai kosong dapat dianggap sebagai tidak ada barang yang terjual, sehingga data masih dapat digunakan untuk analisis. Sementara itu, baris dengan menu kosong dihapus karena nama menu merupakan informasi utama untuk mengidentifikasi data. Jika nama menu tidak diketahui, data tersebut menjadi kurang berguna untuk dianalisis.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

Setelah drop_duplicates(), jumlah baris dataset berkurang sesuai dengan jumlah data yang duplikat. Memastikan tipe data sudah benar penting agar data dapat dianalisis dan dihitung dengan tepat. Misalnya, harga harus bertipe integer agar dapat digunakan dalam operasi matematika.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

Menu yang menghasilkan total_pendapatan tertinggi adalah menu dengan nilai pendapatan paling besar pada hasil groupby. Informasi ini dapat membantu kantin menentukan menu yang perlu lebih banyak disediakan, mempertahankan stok bahan, dan membuat strategi penjualan berdasarkan menu yang memberikan pendapatan lebih besar.
